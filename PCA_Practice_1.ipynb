```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_wine
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

wine = load_wine()
wine_df = pd.DataFrame(wine.data, columns=wine.feature_names)

scaler = StandardScaler()
df = scaler.fit_transform(wine_df)

pca = PCA(n_components=2)
pca_df = pca.fit_transform(df)

plt.figure(figsize=(10, 7))
colors = ['navy','turquoise','darkorange']
lw = 2

for color, i, target_name in zip(colors, [0,1,2], wine.target_names):
    plt.scatter(pca_df[wine.target == i, 0], pca_df[wine.target == i, 1], color=color, alpha=.8, lw=lw, label=target_name)

plt.legend(loc='best', shadow=False, scatterpoints=1)
plt.title('PCA of Wine dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()

print(f'df shape: {df.shape}')
print(f'pca_df shape: {pca_df.shape}')
```

```python
print(pca.explained_variance_ratio_)
```
